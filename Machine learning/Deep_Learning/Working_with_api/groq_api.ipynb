```python
import os
from groq import Groq

GROQ_API_KEY = "your api key"

#get client from api
client = Groq(
    api_key = GROQ_API_KEY
) 

#to get response from server
chat_completion = client.chat.completions.create(
    model = "llama-3.3-70b-versatile",
    messages=[
        {
            "role":"user",
            "content":"Tell me about how beautiful nepal is and what are the things that make nepal beautiful"
        }
    ]
    
)
print(chat_completion.choices[0].message.content)
```

```python
chat = client.chat.completions.create(
    model = "llama-3.3-70b-versatile",
    messages = [
        {
            "role":"user",
            "content":"a+2=7,find a"
        },
        {
            "role": "system",
            "content":"act like a math teacher"
        }
    ]
    
)
print(chat.choices[0].message.content)
```

# Generating Images in openaiapi

### Responsive api

```python
# image => base64 encoded strings => decode

import base64
response = client.response.create(
    model = "gpt-5",
    input = "Generate an image of highly detailed futuristic spaceship,flying near a nebula",
    tools=[{"type":"image_generation"}]
)

image_data = [
    output.result
    for output in response.output
    if output.type == "image_generation_call"
]

#decode and store image in file
if image_data:
    image_base64 = image_data[0]
    with open("spaceship.png","wb") as f:
        image_bytes = base64.b64decode(image_base64)
```
