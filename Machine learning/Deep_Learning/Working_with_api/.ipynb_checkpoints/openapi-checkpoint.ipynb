# Generating image in OPENAPI

### Response api

```python
client = OpenAI(api_key = "OPENAI_API_KEY")

# image => base64 encoded strings => decode

import base64
response = client.responses.create(
    model = "gpt-5",
    input = "Generate an image of highly detailed futuristic spaceship,flying near a nebula",
    tools=[{"type":"image_generation"}]
)

image_data = [
    output.result
    for output in response.output
    if output.type == "image_generation_call"
]

#decode and store image in file
if image_data:
    image_base64 = image_data[0]
    with open("spaceship.png","wb") as f:
        image_bytes = base64.b64decode(image_base64)
```

### Image api

```python
prompt = """Generate an image of highly detailed futuristic spaceship,flying near a nebula"""

result = client.images.generate(
    model = "gpt-image-1.5",
    prompt = prompt
)

image_base64 = result.data[0].b64_json
image_bytes = base64.b64decode(image_base64)

#save the image to file
with open("spaceship-image.png","wb")as f:
    f.write(image_bytes)
```

### edit image

```python
prompt = "Take the reference image and change the spaceship color to red"
result = client.images.edit(
    model = "gpt-image-1.5",
    prompt = prompt,
    image = [
        open("spaceship-image.png","rb"),
    ]
)

image_base64 = result.data[0].b64_json
image_bytes = base64.b64decode(image_base64)

#save the image to file
with open("spaceship-red.png","wb") as f:
    f.write(image_bytes)
```
